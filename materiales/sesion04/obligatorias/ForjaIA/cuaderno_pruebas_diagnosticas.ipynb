# Cuaderno de prácticas · Pruebas diagnósticas
Bioestadística · Grado en Enfermería. Este cuaderno es la idea original de la app.

```python
from math import sqrt

def ic_prop(x, n, z=1.96):
    p = x / n
    centro = (p + z*z/(2*n)) / (1 + z*z/n)
    margen = z * sqrt(p*(1-p)/n + z*z/(4*n*n)) / (1 + z*z/n)
    return p, centro - margen, centro + margen   # Wilson

def prueba_diagnostica(vp, fp, fn, vn):
    se = ic_prop(vp, vp + fn); es = ic_prop(vn, vn + fp)
    vpp = ic_prop(vp, vp + fp); vpn = ic_prop(vn, vn + fn)
    rvp = se[0] / (1 - es[0]); rvn = (1 - se[0]) / es[0]
    return dict(sensibilidad=se, especificidad=es, VPP=vpp, VPN=vpn, RVpos=rvp, RVneg=rvn)
```

```python
r = prueba_diagnostica(vp=34, fp=16, fn=6, vn=144)  # monofilamento y neuropatía
for k, v in r.items():
    print(k, v if isinstance(v, float) else '%.2f (IC 95%%: %.2f a %.2f)' % v)
```

## Interpretación
De cada 100 pacientes con neuropatía, la prueba detecta 85. De cada 100 positivos, 68 tienen neuropatía.
