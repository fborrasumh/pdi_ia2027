Pregunta: ¿difiere la presión arterial sistólica entre fumadores y no fumadores?

```python
import numpy as np
from scipy import stats
rng = np.random.default_rng(7)
fuma = rng.normal(134, 14, 28).round()
no_fuma = rng.normal(126, 13, 32).round()
print('Fumadores: media %.1f DE %.1f n=%d' % (fuma.mean(), fuma.std(ddof=1), len(fuma)))
print('No fumadores: media %.1f DE %.1f n=%d' % (no_fuma.mean(), no_fuma.std(ddof=1), len(no_fuma)))
```

```python
print(stats.shapiro(fuma))
print(stats.shapiro(no_fuma))
print(stats.levene(fuma, no_fuma))
```

```python
t = stats.ttest_ind(fuma, no_fuma, equal_var=True)
print(t)
ci = t.confidence_interval()
print('IC 95%% de la diferencia: %.1f a %.1f mmHg' % (ci.low, ci.high))
```

## Conclusión
Como p < 0,05 rechazamos la hipótesis nula: fumar aumenta la presión arterial sistólica. La diferencia es de unos 8 mmHg.
